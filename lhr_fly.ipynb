```python
def call_payoff(S, K):
    return np.maximum(0.0, S - K)

def put_payoff(S, K):
    return np.maximum(0.0, K - S)

def payoff(S):
    payoff = (
    2 * put_payoff(S, 6200)
    + call_payoff(S, 6200)
    - 2 * call_payoff(S, 6600)
    + 3 * call_payoff(S, 7000)
    )
    return payoff
```
