```python
def call_payoff(S, K):
    return np.maximum(0.0, S - K)

def put_payoff(S, K):
    return np.maximum(0.0, K - S)

def payoff(S):
    payoff = (
    2 * put_payoff(S, 6200)
    + call_payoff(S, 6200)
    - 2 * call_payoff(S, 6600)
    + 3 * call_payoff(S, 7000)
    )
    return payoff
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Payoff helpers
def call_payoff(S, K):
    return np.maximum(0.0, S - K)

def put_payoff(S, K):
    return np.maximum(0.0, K - S)

# Price grid
S = np.linspace(5000, 8800, 2000)

# Structure: +2P6200 + C6200 -2C6600 +3C7000
payoff = (
    2 * put_payoff(S, 6200)
    + call_payoff(S, 6200)
    - 2 * call_payoff(S, 6600)
    + 3 * call_payoff(S, 7000)
)

# Plot
plt.figure(figsize=(9, 5))
plt.plot(S, payoff)
plt.axhline(0)
for K in [6200, 6600, 7000]:
    plt.axvline(K, linestyle="--")
plt.title("Payoff at Expiry: +2P6200 + C6200 -2C6600 +3C7000")
plt.xlabel("ETF Price at Expiry")
plt.ylabel("Payoff (Intrinsic Value Only)")
plt.grid()
plt.tight_layout()
plt.show()
```
